```python
import osmnx as ox
import pandas as pd
import networkx as nx

# 데이터 로드
data = pd.read_csv("../data/node_gangwon_합본.csv")

# '시군'이 '인제'인 데이터만 필터링
inje_data = data[data['시군'] == '인제군']

# 시설명과 노드 ID를 연결
facility_names = inje_data['시설이름']
nodes = inje_data['nearest_node']

# 강원특별자치도의 도로 네트워크 로드
G = ox.graph_from_place('강원특별자치도, 대한민국', network_type='drive')

# 최단 거리 계산
distances = pd.DataFrame(index=facility_names, columns=facility_names)
for i, node1 in enumerate(nodes):
    for j, node2 in enumerate(nodes):
        if node1 != node2:
            try:
                length = nx.shortest_path_length(G, node1, node2, weight='length')
                distances.iloc[i, j] = length
            except nx.NetworkXNoPath:
                distances.iloc[i, j] = None  # 경로가 없는 경우 None으로 설정

# CSV 파일로 저장
distances.to_csv('../data/distances_inje.csv', encoding='CP949')

# 결과 출력
print(distances)
```

```python
import osmnx as ox
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt


G = ox.graph_from_place('Wonju, South Korea', network_type='drive')

data = pd.read_csv("../data/filter_2.csv")  # 파일 변경
df = pd.DataFrame(data)
df.head()
df.set_index('시설명', inplace=True)

# 겹치는 파일명 수정
unique_names = list(dict.fromkeys(df.index))



filtered_df = df.loc[unique_names, unique_names].copy()
for col in filtered_df.columns:
    filtered_df.loc[col, col] = None

min_distance = filtered_df.min().min()
min_distance_idx = filtered_df.stack().idxmin()

print(f"최단 거리: {min_distance}")
print(f"경로: {min_distance_idx[0]} -> {min_distance_idx[1]}")
```

```python
import osmnx as ox
import pandas as pd

# 데이터 로드
data = pd.read_csv('../data/시설 정보.csv')

# 강원특별자치도의 도로 네트워크 그래프 로드
G = ox.graph_from_place('강원특별자치도, 대한민국', network_type='drive')

# 각 좌표에 대해 가장 가까운 노드 번호를 찾기
data['nearest_node'] = data.apply(lambda row: ox.nearest_nodes(G, row['경도'], row['위도']), axis=1)

# 결과 출력
print(data[['위도', '경도', 'nearest_node']])

# CSV 파일로 저장
data.to_csv('../data/시설노드', index=False, encoding='utf-8')
```
